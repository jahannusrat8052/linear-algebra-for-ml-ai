```python
# Jensen's Inequality Demo in Python (Colab Ready)

import numpy as np
import matplotlib.pyplot as plt

# -----------------------
# Jensen's inequality
# Convex function: f(E[X]) <= E[f(X)]
# Concave function: f(E[X]) >= E[f(X)]
# -----------------------

# Sample distribution
np.random.seed(42)
X = np.random.normal(1, 2, 1000)   # Normal distribution

# Convex function: f(x) = x^2
def f_convex(x):
    return x**2

# Concave function: f(x) = sqrt(x), ensure positivity
def f_concave(x):
    return np.sqrt(np.maximum(x, 0))

# ---- Convex case ----
EX = np.mean(X)
lhs_convex = f_convex(EX)      # f(E[X])
rhs_convex = np.mean(f_convex(X))  # E[f(X)]

print("Convex function f(x)=x^2")
print("E[X] =", EX)
print("f(E[X]) =", lhs_convex)
print("E[f(X)] =", rhs_convex)
print("Jensen holds (f(E[X]) <= E[f(X)]):", lhs_convex <= rhs_convex)
print()

# ---- Concave case ----
X_pos = X - np.min(X) + 1e-6   # shift to positive domain
EX_pos = np.mean(X_pos)
lhs_concave = f_concave(EX_pos)
rhs_concave = np.mean(f_concave(X_pos))

print("Concave function f(x)=sqrt(x)")
print("E[X] (shifted) =", EX_pos)
print("f(E[X]) =", lhs_concave)
print("E[f(X)] =", rhs_concave)
print("Jensen holds (f(E[X]) >= E[f(X)]):", lhs_concave >= rhs_concave)

# -----------------------
# Visualization
# -----------------------

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Convex plot
xs = np.linspace(min(X)-1, max(X)+1, 200)
axes[0].plot(xs, f_convex(xs), label="f(x)=x^2")
axes[0].axvline(EX, color="k", linestyle="--", label="E[X]")
axes[0].scatter([EX], [lhs_convex], color="red", zorder=10, label="f(E[X])")
axes[0].scatter(X, f_convex(X), alpha=0.02, label="f(X)")
axes[0].set_title("Convex: f(E[X]) ≤ E[f(X)]")
axes[0].legend()

# Concave plot
xs_pos = np.linspace(min(X_pos), max(X_pos), 200)
axes[1].plot(xs_pos, f_concave(xs_pos), label="f(x)=sqrt(x)")
axes[1].axvline(EX_pos, color="k", linestyle="--", label="E[X]")
axes[1].scatter([EX_pos], [lhs_concave], color="red", zorder=10, label="f(E[X])")
axes[1].scatter(X_pos, f_concave(X_pos), alpha=0.02, label="f(X)")
axes[1].set_title("Concave: f(E[X]) ≥ E[f(X)]")
axes[1].legend()

plt.tight_layout()
plt.show()
```
